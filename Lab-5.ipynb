###Fixing Data Types

```python

print("Before:")
print(clean_df.dtypes)

num_cols = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

for col in num_cols:
    clean_df[col] = pd.to_numeric(
        clean_df[col], errors="coerce"
    )

print("\nAfter:")
print(clean_df.dtypes)

print("\nMissing Values:")
print(clean_df.isnull().sum())
```

###Handling Missing Values

```python


num_cols = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

print("Before:")
print(clean_df.isnull().sum())

for col in num_cols:
    clean_df[col] = clean_df[col].fillna(
        clean_df[col].median()
    )

for col in ["species", "island", "sex"]:
    clean_df[col] = clean_df[col].fillna(
        clean_df[col].mode()[0]
    )

print("\nAfter:")
print(clean_df.isnull().sum())

print("\nDataset Shape:", clean_df.shape)
```

###Handling Duplicates

```python

print("Duplicates Before:")
print(clean_df.duplicated().sum())

clean_df = clean_df.drop_duplicates()

print("\nDuplicates After:")
print(clean_df.duplicated().sum())

print("\nDataset Shape:")
print(clean_df.shape)
```

###Detecting Outliers Using IQR

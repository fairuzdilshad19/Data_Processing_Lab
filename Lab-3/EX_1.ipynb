```python
import requests
import pandas as pd
```

```python
url = (
    "https://api.open-meteo.com/v1/forecast?"
    "latitude=23.8103&longitude=90.4125"
    "&daily=temperature_2m_max,temperature_2m_min,precipitation_sum"
    "&timezone=Asia%2FDhaka&forecast_days=7"
)

weather_response = requests.get(url, timeout=20)
print("HTTP status code:", weather_response.status_code)
weather_response.raise_for_status()
weather_data = weather_response.json()

print("Top-level keys:", weather_data.keys())
```

```python

daily = weather_data["daily"]
df = pd.DataFrame({
    "date": daily["time"],
    "max_temp": daily["temperature_2m_max"],
    "min_temp": daily["temperature_2m_min"],
    "rain": daily["precipitation_sum"]
})

print("First five records:")
display(df.head())
```

```python

print("Dataset size:", df.shape)
print("Column names:", df.columns)

print("Data types:")
print(df.dtypes)

print("Missing values:")
print(df.isnull().sum())
```

```python

df.to_csv("dhaka_weather.csv", index=False)

print("Dataset saved as dhaka_weather.csv")
```
